## Inspect Data for Consistency and Missing Values

### Subtask:
Examine the dataset for any missing values, particularly in critical columns like 'location', 'type of attack', 'device of attack', and 'security type'. We'll also check unique values to ensure data consistency and quality before proceeding with the analysis.


```python
df = load_and_explore_data("/cyber_threats_2020_2024.csv")

print("\n--- Checking Missing Values in Critical Columns ---")
critical_columns = ['Attacker_Location', 'Target_Location', 'Attack_Type', 'Target_Device', 'Security_Type']
for col in critical_columns:
    print(f"Missing values in '{col}': {df[col].isnull().sum()}")

print("\n--- Checking Unique Values in Critical Columns ---")
for col in critical_columns:
    print(f"\nUnique values in '{col}':")
    print(df[col].unique().tolist())
```

## Analyze and Visualize Top Security Types

### Subtask:
Identify the most frequently utilized 'security types' in the dataset. We will then create a bar chart to visually represent the distribution of these security measures, highlighting the most common types of defenses in place.

**Reasoning**:
To identify and visualize the most frequent security types, I will calculate the value counts for the 'Security_Type' column and then create a bar chart, ensuring clear titles and labels. I will also address the previously identified missing values in this column by filling them with 'Unknown' before counting to ensure they are included in the visualization.

**Reasoning**:
Following the previous reasoning, I will now implement the code to fill missing values in 'Security_Type' with 'Unknown', calculate its frequency, and visualize it with a bar chart.



```python
print("\n--- Analyzing and Visualizing Top Security Types ---")

# Fill missing values in 'Security_Type' with 'Unknown'
df['Security_Type_Cleaned'] = df['Security_Type'].fillna('Unknown')

# Calculate the frequency of each unique value in the 'Security_Type_Cleaned' column
security_type_counts = df['Security_Type_Cleaned'].value_counts()

# Display the sorted frequencies
print("\nMost Prevalent Security Types:\n")
print(security_type_counts)

# Create a bar chart to visualize the distribution of security types
plt.figure(figsize=(12, 7))
sns.barplot(x=security_type_counts.index, y=security_type_counts.values, hue=security_type_counts.index, palette='plasma', legend=False)
plt.title('Top Security Types', fontsize=16)
plt.xlabel('Security Type', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
